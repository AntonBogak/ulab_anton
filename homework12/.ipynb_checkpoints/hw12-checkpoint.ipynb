## 2 Plotting
Make a scatter plot of **exoplanet Earth radius vs. equilibrium temperatures**. Include units, x/y axis labels, and a legend.

```python
import matplotlib.pyplot as plt
for index, row in planet_data.iterrows():
    plt.scatter(row['pl_rade'], row['pl_eqt'], label=row['pl_name'])
plt.xlabel('Planet Radius (Earth Radii)')
plt.ylabel('Equilibrium Temperature (Kelvin)')
plt.legend(title='Planets', loc='best')
```

What trend do you notice in the data? Does it have positive or negative correlation?

```python
#The general trend is that planets with larger radii have higher equilibrium temperatures. This is a positive correlation
```
