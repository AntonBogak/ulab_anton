# 2 Plotting a Function
Goal: Write a **function** that will plot $f(x) = x^2$ and $g(x) = \sqrt{x}$ on the same plot and return the $f(x)$ and $g(x)$ values as separate lists. Your plot needs to have the following:
- An appropriate figure size (`figsize=(10,8)` is recommended)
- Labeled x and y axis
- A title
- Different colors and linestyles for each function
- A legend
- Your domain should be $x \in [0, 10]$. *Hint: Use `np.linspace` to set a range of values for your functions.*

Your function should return:
- A list of $f(x)$ values
- A list of $g(x)$ values.

```python
def rootplot():
    xf = np.linspace(1,10,10)
    xg = xf
    yf = [item**2 for item in xf]
    yg = [np.sqrt(item) for item in xf]
    print(yg)
    print(yf)
    plt.figure(figsize=(10,8))
    plt.plot(xf, yf, color='blue', linestyle='--', label='X Squared')
    plt.plot(xg, yg, color='red', linestyle='-.', label='Square Root of X')
    plt.xlabel("X", fontsize=12)
    plt.ylabel("Y", fontsize=12)
    plt.title("Plot of x vs y", fontsize=14)
    plt.legend()

rootplot()
```

Make a **new function** (that has the same foundation as your first function in this question) but includes **all** the changes you described in your comments. Don't forget to call the function and show its output. 

```python
def new_rootplot():
    # Change the domain of x to [0, 20]
    xf = np.linspace(0, 20, 20)  # Generates 20 evenly spaced points from 0 to 20
    yf = [item**2 for item in xf]
    yg = [np.sqrt(item) for item in xf]

    plt.figure(figsize=(10, 8))
    plt.plot(xf, yf, color='blue', linestyle='--', label='X Squared')
    plt.plot(xf, yg, color='red', linestyle=':', label='Square Root of X')

    plt.xlabel("X", fontsize=12)
    plt.ylabel("Y", fontsize=12)
    plt.title("Plot of x vs y", fontsize=14)
    plt.legend(loc='center left', bbox_to_anchor=(1, 0.5))

    # Set x-axis limits to [0, 15] and y-axis limits to [0, 100]
    plt.xlim(0, 15)
    plt.ylim(0, 100)

    # Add a vertical line at x=5
    plt.axvline(x=5, color='green', linestyle='-', linewidth=1.5, label='Vertical Line at x=5')

    # Add a horizontal line at y=25
    plt.axhline(y=25, color='purple', linestyle='--', linewidth=1.5, label='Horizontal Line at y=25')

    # Output the sum of yf and yg values
    sum_yf = sum(yf)
    sum_yg = sum(yg)
    print("Sum of yf (X Squared values):", sum_yf)
    print("Sum of yg (Square Root of X values):", sum_yg)

    plt.legend(loc='center left', bbox_to_anchor=(1, 0.5))
    plt.show()

# Call the function to plot and display the sums
new_rootplot()
```

```python
import numpy as np
import math
import matplotlib.pyplot as plt
def mode_shape(x,y,a,b,m,n):
    return (np.sin((m*math.pi*x)/a))*(np.sin((n*math.pi*y)/b))
# Length
a = 3
# Width
b = 2
# Parameter m
m = 3
# Parameter n
n = 2

# Creates x- and y-arrays
x = np.linspace(0,a,1000)
y = np.linspace(0,b,1000)

# Creates combined X and Y grid
X,Y = np.meshgrid(x,y)

# Computes Z
Z = mode_shape(X,Y,a,b,m,n)

fig_6 = plt.figure()

# Creates axes with 3d projection
ax = plt.axes(projection='3d')

# Plots surface
ax.plot_surface(X, Y, Z, cmap='plasma')

# Sets title
ax.set_title('Natural mode shape for m = 3, n = 2', fontsize='14')

# Sets labels
ax.set_xlabel('X', fontsize = '12')
ax.set_ylabel('Y', fontsize = '12')
ax.set_zlabel('Z', fontsize = '12')

# Sets axes aspect ratio
ax.set_aspect('equal', 'box')

plt.show() 
```
