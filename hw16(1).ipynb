```python
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
# Load Iris dataset
iris = datasets.load_iris()
X = iris.data[:, :2]  # Use first two features: sepal length and sepal width
y = iris.target  # Target is the species 
# Standardize data 
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
# Convert to PyTorch tensors because originally it was upset I was using numpy ndarrays
X_tensor = torch.tensor(X_scaled, dtype=torch.float32)
y_tensor = torch.tensor(y, dtype=torch.long)  #
# Split data into training and testing
X_train, X_test, y_train, y_test = train_test_split(X_tensor, y_tensor, test_size=0.2, random_state=42)
model = torch.nn.Sequential(
    torch.nn.Linear(2, 100),      
    torch.nn.Sigmoid(),          
    torch.nn.Linear(100, 60),     
    torch.nn.Sigmoid(),          
    torch.nn.Linear(60, 3)       # Output layer with 3 classes (species)
)
# Loss function 
loss_fn = torch.nn.CrossEntropyLoss()
# Optimizer
optimizer = torch.optim.Adam(model.parameters(), lr=1e-1)
# Training loop
for t in range(1100):  # Train for 1100 epochs
    y_pred = model(X_train)  
    loss = loss_fn(y_pred, y_train) 

    optimizer.zero_grad()  
    loss.backward()  
    optimizer.step() 

    if t % 100 == 0: 
        print(f"Epoch {t+1}, Loss: {loss.item()}")
# Evaluation
model.eval()
with torch.no_grad():
    output = model(X_test)
    _, predicted = torch.max(output, 1)
    accuracy = (predicted == y_test).sum().item() / y_test.size(0)
    print(f"Test Accuracy: {accuracy * 100:.2f}%")
```
